```python
import warnings
warnings.filterwarnings("ignore")
from astropy.table import Table, Column, vstack
from astropy.coordinates import SkyCoord
from astropy.time import Time
import astropy.units as u
import numpy as np
import pandas as pd
from astropy.io import fits
from utils import *
from catalog import *
from photometry import *
from subtraction import run_hotpants_single
```

```python
sciim = '/home/liangrd/optical_data/EP260131a/NTT/asteris_stacked/0222_r_restored.fits'
refim = '/home/liangrd/optical_data/EP260131a/legacysurvey-1498m032-image-r.fits.fz'
```

```python
#reproject
from reproject import reproject_interp

out_file = "/home/liangrd/optical_data/EP260131a/NTT/0222_r_restored_reprojected.fits"

# -----------------------------
# Read science image
# -----------------------------

sci_hdu = fits.open(sciim)[0]
sci_data = sci_hdu.data
sci_header = sci_hdu.header

# -----------------------------
# Read reference image (target WCS + shape)
# -----------------------------

ref_hdu = fits.open(refim)[0]
ref_header = ref_hdu.header
ref_shape = ref_hdu.data.shape

# -----------------------------
# Reproject science image onto reference grid
# -----------------------------

reprojected_data, footprint = reproject_interp(
    (sci_data, sci_header),
    ref_header,
    shape_out=ref_shape
)

# -----------------------------
# Save result
# -----------------------------

hdu = fits.PrimaryHDU(data=reprojected_data, header=ref_header)
hdu.writeto(out_file, overwrite=True)
print(f"Saved: {out_file}")
```
