```python
#Pre-Processing
import glob, os
from astropy.io import fits
FITS_DIR = "/home/liangrd/optical_data/EP260131a/NTT/"
bands = ['u','g','r']

TELESCOP = 'NTT'
INSTRUME = 'ultracam'

for band in bands:
    flist = glob.glob(os.path.join(FITS_DIR,f'**_{band}_**.fits'))
    for f in flist:
        hdul = fits.open(f)
        header = hdul[0].header
        
        if not header.get('TELESCOP',None):
            header['TELESCOP'] = TELESCOP
        
        if not header.get('INSTRUME',None):
            header['INSTRUME'] = INSTRUME
            
        if not header.get('FILTER',None):
            header['FILTER'] = band
            
        if not header.get('GAIN',None):
            header['GAIN'] = 1.0
        
        hdul[0].header = header
        
        hdul.writeto(f,overwrite=True)
    
    

    
    
```

```python
#!/usr/bin/env python3
"""
Example AutoPhOT driver script.
"""
import os
from autophot import AutomatedPhotometry, prepare_template_directory

# Load default configuration
config = AutomatedPhotometry.load()
config["nCPU"] = 4  # Parallel processing

# Paths
config["outdir_name"] = "REDUCED"
config["wdir"] = "/home/liangrd/optical_data/EP260131a/NTT"
config["fits_dir"] = FITS_DIR

# Target coordinates
config["target_name"] = "EP260131a"
config["target_ra"] = 149.9025
config["target_dec"] = -3.3073

# Photometric catalogs by filter
config["catalog"]["use_catalog"] = {
    "griz": "pan_starrs",
    "u": "gaia", 
    "UBVRI": "apass",
}

# Processing options
config["cosmic_rays"]["remove_cmrays"] = False
config["wcs"]["redo_wcs"] = True

# Enable template subtraction
config["template_subtraction"]["do_subtraction"] = False
# config["template_subtraction"]["method"] = "sfft"
# config["template_subtraction"]["alignment_method"] = "swarp"  # SCAMP+SWarp onto science grid (recommended)

# Create template directories
prepare_template_directory(
    fits_dir=config["fits_dir"],
    include_legacy_p_folders=False,
    confirm_before_continue=True,
)

# Run photometry
output = AutomatedPhotometry.run_photometry(default_input=config, do_photometry=True)

# Generate plots and tables
from lightcurve import plot_lightcurve, generate_photometry_table
plot_lightcurve(output, snr_limit=3, method="PSF")
generate_photometry_table(output, snr_limit=3, method="PSF")
```
