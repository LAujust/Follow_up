```python
#Pre-Processing
import glob, os
from astropy.io import fits
# FITS_DIR = "/home/liangrd/optical_data/EP260131a/NTT/"
# bands = ['u','g','r']
# TELESCOP = 'NTT'
# INSTRUME = 'ultracam'

FITS_DIR = '/home/liangrd/optical_data/EP260518a/TRT'
bands = ['Rc','Ic','V']
TELESCOP = 'TRT'
INSTRUME = 'RED'

for band in bands:
    flist = glob.glob(os.path.join(FITS_DIR,f'**_{band}_**.fits'))
    for f in flist:
        hdul = fits.open(f)
        header = hdul[0].header
        
        if not header.get('TELESCOP',None):
            header['TELESCOP'] = TELESCOP
        
        if not header.get('INSTRUME',None):
            header['INSTRUME'] = INSTRUME
            
        if not header.get('FILTER',None):
            if band in ['Rc','Ic']:
                band = band[0].lower()
            print(band)
            header['FILTER'] = band
        
        if band in ['Rc','Ic']:
            band = band[0].lower()
        print(band)
        header['FILTER'] = band
            
        if not header.get('GAIN',None):
            header['GAIN'] = 1.0
        
        hdul[0].header = header
        
        hdul.writeto(f,overwrite=True)
    
```

```python
import requests
import pandas as pd
import numpy as np
                        
ra = 149.9025
dec = -3.3073
radius_deg = 0.1
                        
# Direct MAST API call for Pan-STARRS
url = url = "https://catalogs.mast.stsci.edu/api/v0.1/panstarrs/dr2/mean"
params = {
    "ra": ra,
    "dec": dec,
    "radius": radius_deg,
    "pagesize": 10000,
    "format": "json"
    }
                        
response = requests.get(url, params=params, timeout=60)
response.raise_for_status()
data = response.json()

                        
if not data:
    electedCatalog = pd.DataFrame()
else:
# Convert to DataFrame, handling None strings
    cols = [data['info'][i]['name'] for i in range(len(data['info']))]
    selectedCatalog = pd.DataFrame(data['data'],columns=cols)

# Replace all forms of None/null with np.nan
    for col in selectedCatalog.columns:
        if selectedCatalog[col].dtype == object:
                                    selectedCatalog[col] = selectedCatalog[col].replace(
                                        ['None', 'none', 'NONE', 'null', 'NULL', 'nan', 'NaN'], np.nan
                                    )
    # Convert numeric columns
    for col in selectedCatalog.columns:
        if col not in ['objName', 'objAltName1', 'objAltName2', 'objAltName3']:
            try:
                selectedCatalog[col] = pd.to_numeric(selectedCatalog[col], errors='coerce')
            except Exception:
                pass                      
                # except Exception as api_exc:
                #         logger.warning(f"Direct Pan-STARRS API failed ({api_exc}), using empty catalog")
                #         selectedCatalog = pd.DataFrame()
                    
# Replace all common null representations
selectedCatalog = selectedCatalog.replace([-999, -999.0, "None", "none", "NONE", "null", "NULL"], np.nan)
columns = [
                        "raMean",
                        "decMean",
                        "raMeanErr",
                        "decMeanErr",
                        "gMeanPSFMag",
                        "gMeanPSFMagErr",
                        "rMeanPSFMag",
                        "rMeanPSFMagErr",
                        "iMeanPSFMag",
                        "iMeanPSFMagErr",
                        "zMeanPSFMag",
                        "zMeanPSFMagErr",
                        "yMeanPSFMag",
                        "yMeanPSFMagErr",
                    ]
selectedCatalog = selectedCatalog[columns]
```

```python
#!/usr/bin/env python3
"""
Example AutoPhOT driver script.
"""
import os
from autophot import AutomatedPhotometry, prepare_template_directory

# Load default configuration
config = AutomatedPhotometry.load()
config["nCPU"] = 4  # Parallel processing

# Paths
config["outdir_name"] = "REDUCED"
config["wdir"] = "/home/liangrd/optical_data/EP260131a/NTT"
config["fits_dir"] = FITS_DIR

# Target coordinates
config["target_name"] = "EP260131a"
config["target_ra"] = 149.9025
config["target_dec"] = -3.3073

# Photometric catalogs by filter
config["catalog"]["use_catalog"] = {
    "griz": "pan_starrs",
    "u": "gaia", 
    "UBVRI": "apass",
}

# Processing options
config["cosmic_rays"]["remove_cmrays"] = False
config["wcs"]["redo_wcs"] = True

# Enable template subtraction
config["template_subtraction"]["do_subtraction"] = False
# config["template_subtraction"]["method"] = "sfft"
# config["template_subtraction"]["alignment_method"] = "swarp"  # SCAMP+SWarp onto science grid (recommended)

# Create template directories
prepare_template_directory(
    fits_dir=config["fits_dir"],
    include_legacy_p_folders=False,
    confirm_before_continue=True,
)

# Run photometry
output = AutomatedPhotometry.run_photometry(default_input=config, do_photometry=True)

# Generate plots and tables
from lightcurve import plot_lightcurve, generate_photometry_table
plot_lightcurve(output, snr_limit=3, method="PSF")
generate_photometry_table(output, snr_limit=3, method="PSF")
```
